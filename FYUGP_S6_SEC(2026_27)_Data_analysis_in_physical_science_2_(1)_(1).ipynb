- Pandas provides two main data structures:
   - **Series**: A one-dimensional labelled array.
   - **DataFrame**: A two-dimensional labelled data structure consisting of rows and columns.

- It supports merging, joining, and concatenating multiple datasets.
- It can perform statistical operations such as calculating mean, median, standard deviation, and correlation.
- It integrates well with NumPy, Matplotlib, and Scikit-learn.

### **Pandas Series**

- A **Pandas Series** is a one-dimensional labelled array used to store data in Python.
- It can store integers, floating-point numbers, strings, and other data types.
- Each element has an associated **index label**, which is used to access the element.
- By default, the index starts from 0.
- A Series can be created using the pd.Series() function.
- It is similar to a single column in an Excel spreadsheet.
- It supports operations such as indexing, filtering, and statistical calculations.
- A Series can be created from a list, tuple, dictionary, or NumPy array.

Example-1

```python
import pandas as pd

# Creating a Series
marks = pd.Series([85, 90, 78, 92, 88])

print(marks)
```

Here, 0, 1, 2, 3, 4 are the index labels, and the numbers are the corresponding values.

Example - 2

You can also create a Series with custom index labels:

```python
marks = pd.Series(
    [85, 90, 78],
    index=['Anu', 'Binu', 'Cathy']
)

print(marks)
```

### **Pandas DataFrame**

- A **Pandas DataFrame** is a two-dimensional labelled data structure used to store and manipulate data in rows and columns.
- It is similar to a table in an Excel spreadsheet or a database table.
- A DataFrame consists of **rows, columns, and index labels**.
- Each column can contain a different data type, such as integers, floating-point numbers, or strings.
- It can be created using the pd.DataFrame() function.
- Data can be imported into a DataFrame from CSV files, Excel files, dictionaries, lists, and NumPy arrays.
- Rows and columns can be accessed using indexing methods such as loc[] and iloc[].
- It supports data cleaning, filtering, sorting, grouping, and statistical analysis.
- Missing values can be identified and handled using functions such as isnull(), dropna(), and fillna().
- DataFrames are widely used in **data science, machine learning, and scientific data analysis**.

Example - 1

A DataFrame is a two-dimensional labelled data structure consisting of rows and columns, similar to a table or an Excel worksheet.

```python
import pandas as pd

# Creating a DataFrame
data = {
    'Name': ['Anu', 'Binu', 'Cathy', 'David'],
    'Physics': [85, 90, 78, 92],
    'Mathematics': [88, 95, 80, 89]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df['Name'])            # Display the Name column
print(df['Physics'].mean())  # Calculate average Physics marks
print(df.iloc[0])            # Display the first row
print(df.head(2))            # Display the first two rows
```

### Difference Between Pandas Series and DataFrame

| Pandas Series | Pandas DataFrame |
|---|---|
| A one-dimensional data structure. | A two-dimensional data structure. |
| Contains a single sequence of values with index labels. | Contains data arranged in rows and columns. |
| Similar to a single column in an Excel sheet. | Similar to a complete table in an Excel sheet. |
| Created using `pd.Series()`. | Created using `pd.DataFrame()`. |
| Has one axis: the index. | Has two axes: rows (index) and columns. |
| Can store different types of data in different Series. | Each column can have a different data type. |
| Example: `pd.Series([10, 20, 30])` | Example: `pd.DataFrame({'A': [10, 20], 'B': [30, 40]})` |
| Access elements using an index, e.g., `s[0]`. | Access columns using column names, e.g., `df['A']`. |
| Represents a single variable or feature. | Represents a dataset containing multiple variables or features. |

# **Create a DataFrame**

A pandas DataFrame can be created by using the pandas.DataFrame function. The syntax of this function can be given as

***pandas.DataFrame(data, index,columns,dtype,copy)***

or

***pd.DataFrame(data, index,columns,dtype,copy)***

where
- **data** can be an ndarray, series,map,list,dictionary,constants or any other data frame
- **index** denotes the row labels for the resulting frame. It is optional argument. By default, if no index is passed, its value will be equal to np.arange(n)
- **columns** is used for specifying column labels.It is optional argument. By default,its value is np.arange(n), if no index is passed.
- **dtype** specifies the data type of each column.
- **copy** is used for copying of data, if the default is false.

## **Creating DataFrame from dictionary**

We can also create a data frame from a dictionary containing ndarrays or lists.But when doing so,remember that all the ndarrys must obe of same length. Moreover, if an index is passed, then the length of the index should be equal to the length of the arrays.In case no index is passed, then by default,index will be range(n), where n is the array length.

### Key points to remeber

- To access a column of this data frame, we just use the notation, *data_frame_object['column_name']*
- To add a column to the data frame,use the notation, *data_frame_object['key']=value*
- To delete a column from the data frame,we can either use the *del* statement by writing del *data_frame_object['column-name']* or using the pop() method by writing *data_frame_object.pop('key')*.
- To access a particular row from the data frame using label or column name, we can use the *loc()* method. It is used by writing *data_frame_object.loc['row-name']*.
- To access a particular row from the data frame, we can even pass an integer location to an *iloc* method by writing *data_frame_object.iloc[row-number]*.
- To access multiple rows from a data frame, use the selection operator (*:*) by writing *data_frame_object.iloc[starting row number: ending row number]*.
- To add new rows at the end of a data frame, use the *append()/concat()* function
- To delete or drop rows from a data frame, the *drop()* method is used. We can specify a row label or index in the *drop()* method. If label is duplicated, then multiple rows will be dropped.

### **Dataframe attributes**

The Pandas DataFrame has a number of column helper functions, which can be used for extrating valuable information from the column. Some of these are



| Method / Attribute | Description |
|---|---|
| `unique()` | This method provides unique elements from a column by removing duplicates. |
| `nunique()` | This method provides number of unique elements from a column by removing duplicates. |
| `mean()` | Returns the mean value of all the items in the column. |
| `head()` | Returns the first n rows. If n is not passed as an argument, then by default, the first five rows are displayed. |
| `tail()` | Returns the last n rows. If n is not passed as an argument, then by default, the last five rows are displayed. |
| `T` | Returns the transpose of a DataFrame by interchanging rows and columns. |
| `axes` | Returns the list of row axis labels and column axis labels. |
| `dtypes` | Returns the data type of each column. |
| `empty` | Returns a Boolean value stating whether the DataFrame object is empty or not. A True value indicates that the object is empty. |
| `ndim` | Returns the number of dimensions of the DataFrame object. |
| `shape` | Returns a tuple representing the dimensionality of the DataFrame. Tuple (a, b), where a represents the number of rows and b represents the number of columns. |
| `size` | Returns the number of elements in the DataFrame. |
| `values` | Returns the data values stored in the DataFrame object as an ndarray. |
